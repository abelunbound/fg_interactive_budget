```python
import pandas as pd
```

```python
# New column names using a dictionary
col_name_dict = {
    "CODE": "line_item_code",
    "LINE ITEM": "line_item_name",
    "AMOUNT": "amount",
    "mda": "mda"
}
input_path = "data/outputs/budget_split_into_capital_and_all_envelopes.xlsx"
sheet_name = "LineItem_Tables"

def ingest_excel(input_path, sheet_name, col_name_dict):

    df = pd.read_excel(input_path, sheet_name=sheet_name)
    df['CODE'] = df['CODE'].astype(str)
    df = df[df['CODE'].str.len() == 8].reset_index(drop=True)
    df = df.rename(columns={df.columns[4]: "mda"})
    df = df.drop(df.columns[3], axis=1)
    df = df.rename(columns=col_name_dict)

    return df

df = ingest_excel(input_path, sheet_name, col_name_dict)
```
