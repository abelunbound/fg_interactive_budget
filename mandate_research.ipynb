# Nigeria Fedaral Government Ministries, Departments and Agencies (MDAs) Mandates

This file researches the official mandate of 820 MDAs that had a budget in 2026 FG budget proposal. The research was done in batches of 20 mdas, mandates provided in .md format and when complete all 800+ were combined in a .csv file to be used for further analysis.

```python
import re
import os
import pandas as pd
import json
```

### Parse single .md file into a dictionary

```python
def parse_mandate_file(filepath):
    """
    Parse a mandate .md file and extract agencies by MDA code.
    Returns a dictionary: {mda_code: mandate_text}
    """
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
    
    # Split by agency entries (using --- separator)
    agencies = re.split(r'\n---\n', content)
    
    mandates_dict = {}
    for agency in agencies:
        # Extract MDA code
        mda_match = re.search(r'\*\*MDA Code:\*\* (\d+)', agency)
        if mda_match:
            mda_code = mda_match.group(1)
            mandates_dict[mda_code] = agency.strip()
    
    return mandates_dict

```

```python
mda_mandates_folder = "data/inputs/fg_agencies_mandate"
mda_mandates_combined_output = "data/outputs/fg_agencies_mandate_output/all_mandates_combined_481_580.md"
```

### Combine multiple batches (.md file dictionaries) into 1 master .md file

```python
def merge_all_mandate_files(folder_path, output_file='combined_mandates.md'):
    """
    Merge all mandate .md files from a folder into one file.
    Sorts by MDA code.
    """
    all_mandates = {}
    
    # Get all .md files
    md_files = [f for f in os.listdir(folder_path) if f.endswith('.md')]
    
    print(f"Found {len(md_files)} .md files")
    
    # Parse each file
    for filename in md_files:
        filepath = os.path.join(folder_path, filename)
        print(f"Processing: {filename}")
        
        mandates = parse_mandate_file(filepath)
        all_mandates.update(mandates)
        print(f"  -> Extracted {len(mandates)} agencies")
    
    print(f"\nTotal agencies: {len(all_mandates)}")
    
    # Sort by MDA code
    sorted_codes = sorted(all_mandates.keys())
    
    # Write combined file
    with open(output_file, 'w', encoding='utf-8') as f:
        f.write("# NIGERIAN FEDERAL AGENCIES MANDATES (COMPLETE)\n\n")
        
        for i, mda_code in enumerate(sorted_codes, 1):
            f.write(all_mandates[mda_code])
            
            # Add separator except for last entry
            if i < len(sorted_codes):
                f.write("\n\n---\n\n")
    
    print(f"\n✅ Combined file saved: {output_file}")
    print(f"   Total agencies: {len(all_mandates)}")
    
    return all_mandates


# # USAGE EXAMPLE
# if __name__ == "__main__":
#     # Option 1: Parse single file
#     mandates = parse_mandate_file('nigeria_agencies_mandates_1_20.md')
#     print(f"Loaded {len(mandates)} agencies")
#     print(f"MDA Codes: {list(mandates.keys())[:5]}...")  # Show first 5
    
#     # Option 2: Merge all files from a folder
#     # merge_all_mandate_files('./mandate_files', 'all_mandates_combined.md')
```

```python
all_mandates = merge_all_mandate_files(mda_mandates_folder, mda_mandates_combined_output)
```

```python
print(all_mandates)
```

```python
# Pretty-print the dictionary
print(json.dumps(all_mandates, indent=4))
```

```python
# Create an empty list to store rows
rows = []

# Process each key-value pair in the dictionary
for mda_code, content in all_mandates.items():
    # Extract the agency name (first line after "##")
    agency_name = content.split("\n\n")[0].replace("## ", "").strip()
    
    # Extract the description (all paragraphs after the second one)
    paragraphs = content.strip().split("\n\n")
    description = "\n\n".join(paragraphs[2:]).strip()
    
    # Append the row as a dictionary
    rows.append({"mda_code": mda_code, "mda_name": agency_name, "mandate_description": description})

# Convert the list of rows into a DataFrame
df = pd.DataFrame(rows)

# Print the DataFrame
print(df)
```
